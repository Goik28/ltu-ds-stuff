Initiate spacy with a small nlp and implement tokenization as asked.

```python
import spacy
import pandas as pd
from collections import Counter

nlp = spacy.load("en_core_web_sm")

testphrase = (
    "This is a test phrase. "
    "its objective is to see the tokenizer, from spacy, in action. "
    "I want to see how it handles: money $4.99 or $25.4; "
    "numbers 32 or 23 and symbols like D&D or acronyms like FBI. "
)

doc = nlp(testphrase)
print("List of tokens: ", list(doc))
# Filter for non-stop, non-punctuation tokens
words = []
for tk in doc:
    if not tk.is_stop and not tk.is_punct:
        words.append(tk.text)
word_freq = Counter(words)
print(word_freq)
print(word_freq.most_common(5))
```

Load the IMDB dataset and explore it.

```python
df = pd.read_csv("imdb_dataset.csv")
print(df.shape)
print(df.columns)
print(df.head())
print(df.tail())
vc = df.value_counts("sentiment")
print(vc)
```

First of all: some clean-up of the reviews to remove HTML tags and extra whitespaces.

```python
print(df["review"].iloc[0])
df["clean_review"] = (
    df["review"]
    .str.replace(r"<[^<>]+>", " ", regex=True)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)
print("-" * 80)
print(df["clean_review"].iloc[0])
print("Number of '<br': ", df["clean_review"].str.contains("<br").sum())
```

Get the first twenty reviews to apply POS tagging.

```python
df_sample = df.head(20).copy()
reviews_docs = nlp.pipe(df_sample["clean_review"].tolist())
for rev_doc in reviews_docs:
    words = []
    for tk in rev_doc:
        words.append([tk.text,tk.pos_,tk.tag_])
    print(words)
```

2. Is there any part of the code that may be optimized (made better)
    >We can disable some of the parts of the pipeline that aren't necessary for the task to get better performance. We execute the same code from the cell above in the cell below with the parser, lemmatizer and entity recognizer disabled and the code executes much quicker.

```python
print(nlp.pipe_names)
df_sample2 = df.head(20).copy()
reviews_docs = nlp.pipe(
    df_sample2["clean_review"].tolist(),
    disable=["parser", "lemmatizer", "ner"])    
for rev_doc in reviews_docs:
    words = []
    for tk in rev_doc:
        words.append([tk.text,tk.pos_,tk.tag_])
    print(words)
```
